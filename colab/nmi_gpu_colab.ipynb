```
# Zip results for download
import zipfile, os
with zipfile.ZipFile('nmi_results.zip', 'w') as zf:
    for f in os.listdir('nmi_results'):
        zf.write(os.path.join('nmi_results', f))
print('Created nmi_results.zip — download from Files panel (left)')
print('\nResults summary:')
import json
for f in sorted(os.listdir('nmi_results')):
    if f.endswith('.json') and f != 'summary.json':
        d = json.load(open(os.path.join('nmi_results', f)))
        b = d.get('baseline', {})
        dpo = d.get('dpo', {})
        print(f"  {d.get('model','?')} s{d.get('seed','?')} {d.get('task','?')}: ASR={b.get('asr',0):.3f}, DPO→{dpo.get('after',{}).get('asr',0):.3f}")
```
